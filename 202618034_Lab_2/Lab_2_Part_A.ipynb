## Part A - Vectorized Programming with NumPy

#### Task 1 - Arrays, Statistics, and Indexing

```python
# Generate Array A with 100 random integers. Set a random seed for reproducibility.
np.random.seed(42)  # Set seed for reproducibility
array_A = np.random.randint(1, 100, size=100)  # Generate 100 random integers between 1 and 100
print("Array A:", array_A)

print('*' * 50)
# Use NumPy functions to find min, max, median, mean, and standard deviation of Array A.
min_a = np.min(np.array_A)
max_a = np.max(np.array_A)
mean_a = np.mean(np.array_A)
median_a = np.median(np.array_A)
std_a = np.std(np.array_A)

print(f"Min: {min_a}, Max: {max_a}, Mean: {mean_a}, Median: {median_a}, Std Dev: {std_a}")
print('*' * 50)

# ● Generate Array B with exactly 100 values using np.arange().
array_b = np.arange(1,101) 
print("Array B:", array_b)
print('*' * 50)

# ● Implement np.zeros() and np.ones(). Display shape and data type.
arr_zeroes = np.zeros((3, 4))  # Create a 3x4 array of zeros
arr_ones = np.ones((2, 5))    # Create a 2x5 array of ones
print("Array of Zeros:\n", arr_zeroes, "\nShape:", arr_zeroes.shape, "Data Type:", arr_zeroes.dtype)
print("Array of Ones:\n", arr_ones, "\nShape:", arr_ones.shape, "Data Type:", arr_ones.dtype)
print('*' * 50)

# ● Implement np.linspace(). Briefly state how it differs from np.arange().
arr_linspace = np.linspace(0,1, num = 5) # Create an array of 5 evenly spaced values between 0 and 1
print(arr_linspace)
print('*' * 50)

# linspace differs from arange by even spacing of values, while arange produces values 
# that lie in a sequence with a specfic step size.

# ● Create one 2D and one 3D array. Show shape, ndim, indexing, rows, columns, and slicing.
arr_2D = np.array([[1,2,3], [4,5,6]])
arr_3d = np.arange(1,25).reshape(3, 4, 2)  # Create a 3D array with shape (3, 4, 2)
print('2-D array :' ,arr_2D) 
print('3-D array :' ,arr_3d) 
print('Shape of 2-D array :' ,arr_2D.shape)
print('Shape of 3-D array :' ,arr_3d.shape)
print('Number of dimensions of 2-D array :' ,arr_2D.ndim)
print('Number of dimensions of 3-D array :' ,arr_3d.ndim)     
print('Indexing 2-D array :' ,arr_2D[0,1]) # Accessing element at first row, second column
print('Indexing 3-D array :' ,arr_3d[1,2,0]) # Accessing element at second block, third row, first column
print('Slicing 2-D array :' ,arr_2D[:,1]) # Accessing all rows, second column
print('Slicing 3-D array :' ,arr_3d[0,:,:]) #


# ● Use reshape() to create a matrix, then flatten it back to 1D.
matrix = np.arange(1, 13).reshape(3, 4)
flattened = matrix.flatten()
print('Original Matrix:\n', matrix)
print('Flattened Array:', flattened)

print('*' * 50)
```

#### Task 2 - Vectorized Arithmetic and Linear Algebra

```python
# Create two matrices. Perform addition, element-wise multiplication, 
# and matrix multiplication using @ or np.matmul().
mat_A = np.array([[1, 2], [3, 4]])
mat_B = np.array([[5, 6], [7, 8]])
print("Matrix A:\n", mat_A)
print("Matrix B:\n", mat_B)
print("Addition (mat_A + mat_B): " , mat_A + mat_B)
print("Element-wise Multiplication (mat_A * mat_B): " , mat_A * mat_B)
print("Matrix Multiplication (mat_A @ mat_B): " , mat_A @ mat_B)

print('*' * 50)

# ● For a square matrix, find its transpose and determinant.
print("Transpose of Matrix A:\n", mat_A.T)
print("Determinant of Matrix A:\n", np.linalg.det(mat_A))

print('*' * 50)

# ● If invertible, find the inverse and verify A @ A^-1 using np.allclose().
print("Inverse of Matrix A:\n", np.linalg.inv(mat_A))
print("Verification (A @ A^-1):\n", np.allclose(mat_A @ np.linalg.inv(mat_A), np.eye(2)))

print('*' * 50)

```

#### Task 3 - Normal Distribution and Histogram


```python
# ● Generate at least 1,000 values from a normal distribution. 
# State the chosen mean and standard deviation.
import numpy as np

mean = 50
std_dev = 10
n = 1000

data = np.random.normal(loc=mean, scale=std_dev, size=n)

print("mean : " , mean)
print("standard deviation : " , std_dev) 
print('*' * 50)

# ● Find the sample mean and sample standard deviation. 
# Compare them with the chosen values.
sample_mean = np.mean(data)
sample_std = np.std(data)

print("Chosen mean:", 50)
print("Chosen standard deviation:", 10)
print("Sample mean:", sample_mean)
print("Sample standard deviation:", sample_std)
print('*' * 50)

# ● Plot a histogram with it.
plt.hist(data, bins=30)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Normally Distributed Data")

plt.show()
```
